```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine
```

```python
from sklearn.datasets import load_wine # Mtulti class dataset

wine_data = load_wine(as_frame= True) # function for loading the dataset
print(type(wine_data))
```

### What do we have in the dataset?

data : {ndarray, dataframe} of shape (178, 13)
            The data matrix. If `as_frame=True`, `data` will be a pandas
            DataFrame.

target: {ndarray, Series} of shape (178,)
            The classification target. If `as_frame=True`, `target` will be
            a pandas Series.
  
feature_names: list
            The names of the dataset columns.
        
target_names: list
            The names of target classes.

frame: DataFrame of shape (178, 14)
            Only present when `as_frame=True`. DataFrame with `data` and
            `target`.

```python
# methods on sklearn datasets
wine_df = wine_data.frame

wine_df.head(10)
```

### Now after we have some idea about the data we will try to train models on it

```python
from sklearn.model_selection import train_test_split

# first we need to divide between X, y and train test

y = wine_df['target'].copy()
X = wine_df.drop(columns = ['target']).copy()
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.05 ,random_state=42)

#split = int(len(wine_df)* 0.8)
#X_train ,y_train = X[ : split] , y[ : split]
#X_test , y_test = X[split : ] , y[split : ]



```

```python
# first model
from sklearn.linear_model import RidgeClassifier

# calling the model with the standard settings
ridge_model = RidgeClassifier()
# training/ fitting the data in to the model
ridge_model.fit(X_train, y_train)
# calculating the predictions for the test set
y_pred = ridge_model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, y_pred)
```
